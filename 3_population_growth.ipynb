```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
po = int(input('Populasi awal (Po): '))
px = int(input('Populasi pada waktu t (Pt): '))
to = int(input('Tahun awal (to): '))
tx = int(input('Tahun pada waktu tx (tx): '))
K = int(input('Kapastas penampung (K): '))
```

```python
A = (K - po) / po
r = - 1 / (tx - to) * (np.log((K / (A * px)) - (1 / A)))

def logistic_model(t, r, K, A):
    return K / (1 + A*np.exp(-r*(t - to)))
```

```python
years_input = input('Masukkan tahun (Pisah dengan koma): ')
years = [int(year) for year in years_input.split(',')]

population = [logistic_model(year, r, K, A) for year in years]

for dYear, dPop in zip(years, population):
    print(f'Populasi pada tahun {dYear}: {int(dPop)} jiwa')
```

```python
if len(years) == 1:
    year_range = np.arange(years[0] - 5, years[0] + 6, 1)
else:
    year_range = np.arange(min(years) - 10, max(years) + 10, 1)
    
population_value = logistic_model(year_range, r, K, A)
```

```python
plt.figure(figsize=(8, 6))
plt.plot(year_range, population_value, label='Logistic Model Population')
plt.scatter(years, population, color='black', label='Prediction Point', zorder=5)
plt.axhline(y=K, color='red', linestyle='--', label='Carrying Capacity (K)')
plt.title('Growth Population')
plt.xlabel('Year')
plt.ylabel('Popluation')

plt.legend()
plt.grid(True)
plt.show()
```
