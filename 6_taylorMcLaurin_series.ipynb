```python
import numpy as np 
import matplotlib.pyplot as plt 
import sympy as sp 
from sympy import series, Symbol, lambdify 
from ipywidgets import interact, FloatSlider, IntSlider
```

```python
x = sp.Symbol('x') 
 
def taylor_series_plotter(): 
    func_str = input("Masukkan fungsi f(x), contoh: sp.sin(x), sp.exp(x), sp.log(1+x): ") 
     
    try: 
        f = eval(func_str) 
    except: 
        print("Error: Fungsi tidak valid!") 
        return 
     
    @interact( 
        titik_ekspansi=FloatSlider(value=0.0, min=-5.0, max=5.0, step=0.5, description='Titik Ekspansi:'), 
        jumlah_suku=IntSlider(value=3, min=1, max=10, step=1, description='Jumlah Suku:'), 
        x_min=FloatSlider(value=-10.0, min=-20.0, max=0.0, step=0.5, description='x min:'), 
        x_max=FloatSlider(value=10.0, min=0.0, max=20.0, step=0.5, description='x max:') 
    ) 
    def plot_taylor(titik_ekspansi, jumlah_suku, x_min, x_max): 
        taylor_series = series(f, x, titik_ekspansi, jumlah_suku + 1).removeO() 
         
        f_num = lambdify(x, f, 'numpy') 
        taylor_num = lambdify(x, taylor_series, 'numpy') 
         
        x_vals = np.linspace(x_min, x_max, 1000) 
         
        try: 
            f_vals = f_num(x_vals) 
            taylor_vals = taylor_num(x_vals) 
        except: 
            print("Error: Domain fungsi tidak valid untuk range yang dipilih!") 
            return 
         
        plt.figure(figsize=(12, 8)) 
        plt.plot(x_vals, f_vals, 'b-', linewidth=2, label='Fungsi Asli') 
        plt.plot(x_vals, taylor_vals, 'r--', linewidth=2, label=f'Taylor (n={jumlah_suku})') 
        plt.axvline(x=titik_ekspansi, color='g', linestyle=':', alpha=0.7, label=f'Titik Ekspansi: {titik_ekspansi}') 
         
        plt.title(f'Deret Taylor dari f(x) = {sp.pretty(f)}') 
        plt.xlabel('x') 
        plt.ylabel('f(x)') 
        plt.legend() 
        plt.grid(True, alpha=0.3) 
        plt.xlim(x_min, x_max) 
         
        print(f"Deret Taylor dengan {jumlah_suku} suku di x = {titik_ekspansi}:") 
        print(sp.pretty(taylor_series)) 
        plt.show()
```

```python
def contoh_fungsi(): 
    print("Useable Function:") 
    print("\n" + "=" * 60)
    
    # TRIGONOMETRI
    print("\n TRIGONOMETRI:")
    print("   • sp.sin(x)          - Fungsi sinus")
    print("   • sp.cos(x)          - Fungsi cosinus") 
    print("   • sp.tan(x)          - Fungsi tangen")
    print("   • sp.atan(x)         - Fungsi arctan")
    print("   • 1/sp.cos(x)        - Fungsi secan")
    print("   • 1/sp.sin(x)        - Fungsi cosecan")
    
    # HIPERBOLIK
    print("\n HIPERBOLIK:")
    print("   • sp.sinh(x)         - Sinus hiperbolik")
    print("   • sp.cosh(x)         - Cosinus hiperbolik")
    print("   • sp.tanh(x)         - Tangen hiperbolik")
    
    # EKSPONENSIAL & LOGARITMA
    print("\n EKSPONENSIAL & LOGARITMA:")
    print("   • sp.exp(x)          - Fungsi eksponensial e^x")
    print("   • sp.exp(-x)         - Fungsi e^(-x)")
    print("   • sp.log(x)          - Logaritma natural ln(x)")
    print("   • sp.log(1+x)        - ln(1+x)")
    print("   • sp.log(1-x)        - ln(1-x)")
    
    # FUNGSI RASIONAL
    print("\n FUNGSI RASIONAL:")
    print("   • 1/x                - Fungsi 1/x")
    print("   • 1/(1+x)            - Fungsi 1/(1+x)")
    print("   • 1/(1-x)            - Fungsi 1/(1-x)")
    print("   • 1/(1+x**2)         - Fungsi 1/(1+x²)")
    print("   • x/(1+x**2)         - Fungsi x/(1+x²)")
    
    # POLINOMIAL
    print("\n POLINOMIAL:")
    print("   • x**2 + 2*x + 1           - Polinomial kuadrat")
    print("   • x**3 - 3*x**2 + 3*x - 1  - Polinomial kubik")
    print("   • 1 + x + x**2             - Polinomial sederhana")
    
    # KOMBINASI FUNGSI
    print("\n KOMBINASI FUNGSI:")
    print("   • sp.sin(x**2)        - sin(x²)")
    print("   • sp.cos(x**2)        - cos(x²)")
    print("   • sp.exp(x)*sp.sin(x) - e^x × sin(x)")
    print("   • sp.exp(x)*sp.cos(x) - e^x × cos(x)")
    print("   • x*sp.sin(x)         - x × sin(x)")
    print("   • x*sp.exp(x)         - x × e^x")
    
    # FUNGSI KHUSUS
    print("\n FUNGSI KHUSUS:")
    print("   • sp.exp(-x**2)      - Gaussian")
    print("   • 1/(1+sp.exp(-x))   - Fungsi sigmoid/logistic")
    print("   • sp.atan(x)         - Fungsi arctan")
    print("   • sp.sin(x)/x        - sinc function")
    
    print("\n" + "=" * 60)
    print("\nGunakan 'sp.' sebelum fungsi matematika (sin, cos, exp, dll)")
```

```python
if __name__ == "__main__": 
    print("Taylor and McLaurin Series Plotter") 
    print("=" * 50) 
    contoh_fungsi() 
    print("\n" + "=" * 50) 
    taylor_series_plotter()
```
